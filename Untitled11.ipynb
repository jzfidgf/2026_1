```python
bag=set()
print("初始背包",bag)
bag.add("劍")
bag.add("盾")
bag.add("藥水")
print("撿到道具後的背包:",bag)
bag.add("藥水")
print("撿到藥水後的背包:",bag)
bag.remove("盾")
print("使用盾後的背包:",bag)
lose=bag.pop()
print("掉落的道具:",lose)
print("遊戲結束，最後背包:",bag)
```

```python
def add(x,y):
  try:
    return x/y
  except ZeroDivisionError:
    return "除數不可為0"
print(add(10,0))
```

```python
try:
  account=input("請輸入帳號:")
  password=input("請輸入密碼:")
  if account == "" or password == "":
    raise ValueError("帳號或密碼不可以為空")
except ValueError:
  print ("帳號密碼不可為空")
else:
  if account =="admin" and password=="1234":
    print("登入成功")
  else:
    print("登入失敗")
finally:
  print("系統結束")
```
