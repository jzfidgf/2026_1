```python
def shopping(p,q):
 total=p*q
 tax=total*0.05
 all=total+tax
 return total, tax, all
a=float(input("請輸入單價: "))
b=float(input("請輸入數量: "))
s_total,s_tax,s_all=shopping(a,b)
print(f"小計:{s_total}")
print(f"稅金:{s_tax}")
print(f"總金額:{s_all}")
```

```python
def f(n):
    if n == 1:
        return 1
    elif n == 2:
        return 2
    else:
        return f(n - 1) + f(n - 2)

print("總共有", f(10), "種方式")
```
